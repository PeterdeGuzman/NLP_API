```python
import pandas as pd
import os
import random

# Define the folder containing the CSV files
DATA_FOLDER = 'math_data'
# Define the name for the output file
OUTPUT_FILE = 'math_data_summarized.csv'
# Define the number of rows to sample from each file
SAMPLE_SIZE = 20

# Initialize an empty list to store the sampled DataFrames
sampled_dataframes = []

# --- 1. Iterate through files and sample data ---
print(f"Starting data sampling from files in: **{DATA_FOLDER}**")

# Use os.listdir to get a list of all files/folders in the specified directory
for filename in os.listdir(DATA_FOLDER):
    
    # 1.1. Check if the file is a CSV and contains 'train'
    # We use .endswith('.csv') and 'train' in filename.lower() for case-insensitive matching
    if filename.endswith('.csv') and 'train' in filename.lower():
        
        # Construct the full file path
        filepath = os.path.join(DATA_FOLDER, filename)
        
        try:
            # Read the CSV file into a DataFrame
            df = pd.read_csv(filepath)
            
            # Ensure the file has enough rows to sample from
            if len(df) >= SAMPLE_SIZE:
                
                # Randomly sample the specified number of rows (SAMPLE_SIZE)
                sampled_df = df.sample(n=SAMPLE_SIZE, random_state=42)
                
                # Add a column to identify the source file (optional but useful)
                sampled_df['source_file'] = filename
                
                # Append the sampled DataFrame to the list
                sampled_dataframes.append(sampled_df)
                print(f"✅ Sampled {SAMPLE_SIZE} rows from: {filename}")
                
            else:
                print(f"⚠️ Skipping {filename}: Only has {len(df)} rows (less than required {SAMPLE_SIZE}).")
                
        except Exception as e:
            print(f"❌ Error reading or processing file {filename}: {e}")

# --- 2. Combine and save the summarized data ---
if sampled_dataframes:
    
    # Concatenate all DataFrames in the list into a single DataFrame
    combined_df = pd.concat(sampled_dataframes, ignore_index=True)
    
    # Save the combined DataFrame to a new CSV file in the current directory
    combined_df.to_csv(OUTPUT_FILE, index=False)
    
    print("\n--- Summary ---")
    print(f"Successfully combined a total of **{len(combined_df)}** rows.")
    print(f"Output saved to: **{OUTPUT_FILE}**")
    
else:
    print("\n--- Summary ---")
    print("No qualifying 'train' CSV files were found in the 'math_data' folder.")
```
