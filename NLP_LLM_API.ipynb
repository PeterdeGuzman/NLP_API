This tool aims to use Google's Gemini API to return the correct answer to a provided math problem. 

```python
from google import genai

client = genai.Client()

response = client.models.generate_content(
    model="gemini-2.5-flash", contents="Explain how AI works in a few words"
)
print(response.text)
```

```python
#setting up argparse

import argparse

parser = argparse.ArgumentParser(description="A simple command-line tool to help solve math problems.")

parser.add_argument("prob", help="Please provide a math problem for Gemini to solve.")

args = parser.parse_args()
```
